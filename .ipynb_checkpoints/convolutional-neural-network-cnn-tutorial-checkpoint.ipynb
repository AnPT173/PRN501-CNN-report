# Convolutional Neural Networks (CNN)

Phân loại ảnh biểu tượng giao thông 28×28 (6 lớp) bằng CNN trên Keras.

**Quy trình**
1. Load dữ liệu
2. Chuẩn hóa, reshape, one-hot encoding
3. Tách train / validation
4. Xây dựng mô hình CNN
5. Compile, tăng cường dữ liệu và huấn luyện
6. Đánh giá

```python
import os
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from keras.layers import Conv2D, Dense, Dropout, Flatten, MaxPool2D
from keras.models import Sequential
from keras.optimizers import Adam
from keras.utils import to_categorical
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator

warnings.filterwarnings("ignore")

DATA_DIR = Path("datasets/transportation_cnn")
if not DATA_DIR.exists():
    DATA_DIR = Path("../datasets/transportation_cnn")
if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Không tìm thấy {DATA_DIR.resolve()}. Chạy notebook từ week2 hoặc PRN501-CNN-report."
    )

print("Dataset files:", sorted(os.listdir(DATA_DIR)))
```

## 1. Load dữ liệu

Mỗi dòng là một ảnh xám 28×28 (784 pixel) kèm nhãn `label` (0–5).

```python
train = pd.read_csv(DATA_DIR / "transportation_train.csv")
test = pd.read_csv(DATA_DIR / "transportation_test.csv")

Y_train = train["label"]
X_train = train.drop(columns=["label"])
Y_test = test["label"]
X_test = test.drop(columns=["label"])

print("train:", train.shape, "| test:", test.shape)
print(Y_train.value_counts().sort_index())
train.head()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(x=Y_train, palette="icefire")
plt.title("Số mẫu theo lớp (train)")
plt.show()

img = X_train.iloc[100].to_numpy().reshape(28, 28)
plt.imshow(img, cmap="gray")
plt.title(f"Mẫu label = {Y_train.iloc[0]}")
plt.axis("off")
plt.show()
```

## 2. Chuẩn hóa, reshape, one-hot encoding

- Chia pixel cho 255 để CNN hội tụ nhanh hơn.
- Đưa ảnh về `(28, 28, 1)` — Keras cần chiều kênh (ảnh xám = 1 kênh).
- One-hot nhãn: ví dụ `2 → [0, 0, 1, 0, 0, 0]`.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0

X_train = X_train.values.reshape(-1, 28, 28, 1)
X_test = X_test.values.reshape(-1, 28, 28, 1)

Y_train = to_categorical(Y_train, num_classes=6)
Y_test = to_categorical(Y_test, num_classes=6)

print("X_train:", X_train.shape, "| Y_train:", Y_train.shape)
print("X_test :", X_test.shape, "| Y_test :", Y_test.shape)
```

## 3. Tách train / validation

Giữ 10% tập train làm validation khi huấn luyện.

```python
X_train, X_val, Y_train, Y_val = train_test_split(
    X_train, Y_train, test_size=0.1, random_state=2
)

print("X_train:", X_train.shape, "| X_val:", X_val.shape)
print("Y_train:", Y_train.shape, "| Y_val:", Y_val.shape)
```

## 4. Xây dựng mô hình CNN

Luồng CNN: **Conv2D → MaxPool → Dropout → Conv2D → MaxPool → Dropout → Flatten → Dense → Softmax**.

- **Conv2D**: lọc đặc trưng (cạnh, hình khối).
- **Same padding**: giữ kích thước feature map.
- **MaxPool**: giảm kích thước, giảm overfitting.
- **Flatten + Dense**: đưa đặc trưng vào lớp fully connected để phân loại 6 lớp.

```python
model = Sequential([
    Conv2D(filters=8, kernel_size=(5, 5), padding="Same", activation="relu", input_shape=(28, 28, 1)),
    MaxPool2D(pool_size=(2, 2)),
    Dropout(0.25),
    Conv2D(filters=16, kernel_size=(3, 3), padding="Same", activation="relu"),
    MaxPool2D(pool_size=(2, 2), strides=(2, 2)),
    Dropout(0.25),
    Flatten(),
    Dense(256, activation="relu"),
    Dropout(0.5),
    Dense(6, activation="softmax"),
])

model.summary()
```

## 5. Compile, data augmentation và huấn luyện

- Optimizer **Adam**, loss **categorical_crossentropy** (đa lớp).
- Data augmentation xoay / zoom / dịch nhẹ để giảm overfitting.

```python
# ==============================================================================
# 3: COMPILE, DATA AUGMENTATION VÀ HUẤN LUYỆN - an26ms23396@fsb.edu.vn
# ==============================================================================

# SỬ DỤNG THUẬT TOÁN ADAM ĐỂ CẬP NHẬT OVERWEIGHT CỦA NEURAL NETWORK
# SỬ DỤNG DATA AUGMENTATION ĐỂ TẠO RA CÁC BIẾN THỂ NHẸ HƠN CỦA ẢNH TRAINING NHẰM GIẢM OVERWRITING

# 3.1: Chọn optimizer Adam với learning rate = 0.001 là mức độ của bước cập nhật weight (Quá lớn => update mạnh, khó hội tụ. Quá nhỏ => update nhỏ, học chậm)  
optimizer = Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.999)
# 3.2: Sử dụng 3 thành phần: optimizer Adam, categorical crossentropy làm hàm tính toán loss, và metric để đếm tỉ lệ dự đoán đúng
model.compile(optimizer=optimizer, loss="categorical_crossentropy", metrics=["accuracy"])
# Do label đã được one-hot ở phía trên nên dùng categorical crossentropy

# Tham số training
epochs = 10 # số lần model đi qua các batch của training data 1 vòng
batch_size = 250 # số lượng data ảnh dược dùng để xử lý

# 3.3: Tiến hành data augmentation để tạo các biến thể nhẹ của ảnh training
datagen = ImageDataGenerator(
    rotation_range=5, # xoay tối đa 5 độ
    zoom_range=0.1, # zoom tối đa 
    width_shift_range=0.1, # dịch theo chiều ngang
    height_shift_range=0.1, # dịch theo chiều dọc
    horizontal_flip=False, 
    vertical_flip=False,
    # do đối tượng là biển báo giao thông, khi flip ngang/dọc sẽ gây sai lệch hoặc có thể biến đổi ý nghĩa của biển báo thành 1 biển báo khác
)

# 3.4: Chuẩn bị generator để xử lý dữ liệu dựa trên training data 
datagen.fit(X_train)

# 3.5: Huấn luyện mô hình bằng model.fit; biến history lưu lịch sử loss/accuracy mỗi epoch (dùng vẽ đồ thị sau này)
history = model.fit(
    # Generator: mỗi batch lấy ảnh từ X_train, nhãn từ Y_train, áp augmentation ngẫu nhiên theo cấu hình datagen
    datagen.flow(X_train, Y_train, batch_size=batch_size),
    # Số epoch: lặp toàn bộ quy trình trên generator đúng `epochs` lần (đang set là 10)
    epochs=epochs,
    # Sau mỗi epoch, đánh giá trên tập validation (không augmentation) để theo dõi val_loss / val_accuracy
    validation_data=(X_val, Y_val),
    # Số batch mỗi epoch = số mẫu train // batch_size = 432 // 250 = 1 
    steps_per_epoch=X_train.shape[0] // batch_size,
)
```

## 6. Đánh giá

Theo dõi quá trình huấn luyện, độ chính xác tổng thể, báo cáo phân loại theo lớp (precision / recall / F1), ma trận nhầm lẫn và minh họa dự đoán trên tập validation và test.

```python
# ==============================================================================
# 4: DỰ ĐOÁN VÀ ĐÁNH GIÁ MÔ HÌNH (PREDICTION & EVALUATION) - huy26ms23394@fsb.edu.vn
# ==============================================================================

# 1. Trực quan hóa Validation Loss qua từng Epoch
# Giúp theo dõi quá trình học của mô hình và phát hiện dấu hiệu Overfitting (nếu val_loss bắt đầu tăng lại)
plt.plot(history.history["val_loss"], color="b", label="validation loss")
plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()

# 2. Thực hiện dự đoán trên tập dữ liệu Validation (X_val)
# Mô hình trả về một mảng chứa 6 giá trị xác suất (Probability) tương ứng với 6 lớp cho mỗi ảnh
Y_pred = model.predict(X_val)

# 3. Sử dụng hàm argmax để lấy chỉ số (index) của lớp có xác suất dự đoán lớn nhất
Y_pred_classes = np.argmax(Y_pred, axis=1)

# 4. Chuyển nhãn thực tế từ dạng One-hot Encoding về lại dạng chỉ số lớp (Integer Class)
Y_true = np.argmax(Y_val, axis=1)

# 5. Tạo ma trận nhầm lẫn (Confusion Matrix) để so sánh các cặp lớp thực tế (Y_true) và lớp dự đoán (Y_pred_classes)
confusion_mtx = confusion_matrix(Y_true, Y_pred_classes)

# 6. Trực quan hóa ma trận nhầm lẫn dưới dạng biểu đồ nhiệt (Heatmap)
plt.figure(figsize=(8, 8))
sns.heatmap(
    confusion_mtx,
    annot=True,         # Hiển thị số lượng mẫu chi tiết ngay trên từng ô của ma trận
    fmt="d",            # Định dạng kiểu hiển thị số nguyên
    cmap="Greens",      # Tông màu hiển thị xanh lá
    linewidths=0.01,    # Độ rộng đường ranh giới giữa các ô
    linecolor="gray",   # Màu đường ranh giới giữa các ô
)
plt.xlabel("Predicted") # Trục hoành (X-axis) biểu thị lớp mô hình dự đoán
plt.ylabel("True")      # Trục tung (Y-axis) biểu thị lớp thực tế (Ground Truth)
plt.title("Confusion Matrix (validation)")
plt.show()

# 8. Đánh giá trên tập test (chưa dùng khi huấn luyện)
# Tập test tách biệt khỏi train/validation — dùng để đo khả năng tổng quát hóa của mô hình trên dữ liệu chưa thấy khi huấn luyện

# 8.1. Tính loss (cross-entropy) và accuracy trên toàn bộ tập test
# Y_test phải cùng định dạng one-hot (6 lớp) như Y_train khi compile model với categorical_crossentropy
test_loss, test_acc = model.evaluate(X_test, Y_test, verbose=0)

# 8.2. Dự đoán phân phối xác suất 6 lớp cho mỗi ảnh trong X_test
Y_test_pred = model.predict(X_test, verbose=0)

# 8.3. Lấy lớp dự đoán: chỉ số có xác suất cao nhất (argmax theo trục lớp)
Y_test_pred_classes = np.argmax(Y_test_pred, axis=1)

# 8.4. Chuyển nhãn thực tế từ one-hot về chỉ số lớp nguyên (0–5) để so sánh với dự đoán
Y_test_true = np.argmax(Y_test, axis=1)

# 8.7. Ma trận nhầm lẫn trên test — hàng = lớp thực, cột = lớp dự đoán
confusion_mtx_test = confusion_matrix(Y_test_true, Y_test_pred_classes)

# 8.8. Trực quan hóa ma trận nhầm lẫn test bằng heatmap (tông xanh dương)
plt.figure(figsize=(8, 8))
sns.heatmap(
    confusion_mtx_test,
    annot=True,         # Hiển thị số lượng mẫu trên từng ô
    fmt="d",            # Định dạng số nguyên
    cmap="Blues",       # Tông màu xanh dương (phân biệt với validation dùng Greens)
    linewidths=0.01,
    linecolor="gray",
)
plt.xlabel("Predicted")  # Lớp mô hình dự đoán
plt.ylabel("True")       # Lớp thực tế (ground truth)
plt.title("Confusion Matrix (test)")
plt.show()

# 9. Minh họa một số dự đoán đúng / sai trên tập test

# 9.1. Đọc file ánh xạ mã lớp (0–5) sang tên lớp (car, bus, ...) để hiển thị dễ đọc
labels_path = DATA_DIR / "class_labels.csv"
if labels_path.exists():
    class_names = pd.read_csv(labels_path)
    name_col = "name" if "name" in class_names.columns else class_names.columns[-1]
    id_col = "label" if "label" in class_names.columns else class_names.columns[0]
    label_to_name = dict(zip(class_names[id_col], class_names[name_col]))
else:
    label_to_name = {i: str(i) for i in range(6)}

# 9.2. Hàm vẽ lưới ảnh grayscale 28×28 kèm nhãn True (T) và Predicted (P)
def show_predictions(images, y_true, y_pred, title, n=12):
    n = min(n, len(images))
    cols = 6
    rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 2, rows * 2))
    axes = np.atleast_1d(axes).flatten()
    for i in range(n):
        axes[i].imshow(images[i].reshape(28, 28), cmap="gray")
        t_name = label_to_name.get(y_true[i], y_true[i])
        p_name = label_to_name.get(y_pred[i], y_pred[i])
        color = "green" if y_true[i] == y_pred[i] else "red"  # Xanh: đúng; Đỏ: sai
        axes[i].set_title(f"T:{t_name}\nP:{p_name}", fontsize=8, color=color)
        axes[i].axis("off")
    for j in range(n, len(axes)):
        axes[j].axis("off")
    fig.suptitle(title, fontsize=12)
    plt.tight_layout()
    plt.show()

# 9.3. Tách chỉ số mẫu dự đoán đúng và sai để lấy mẫu minh họa
correct_idx = np.where(Y_test_true == Y_test_pred_classes)[0]
wrong_idx = np.where(Y_test_true != Y_test_pred_classes)[0]

# 9.4. Hiển thị tối đa 12 ảnh dự đoán đúng (nếu có) — kiểm tra mô hình học tốt các lớp nào
if len(correct_idx):
    pick = correct_idx[:12]
    show_predictions(X_test[pick], Y_test_true[pick], Y_test_pred_classes[pick], "Dự đoán đúng (test)")

# 9.5. Hiển thị tối đa 12 ảnh dự đoán sai (nếu có) — phân tích lỗi nhầm lẫn giữa các lớp
if len(wrong_idx):
    pick = wrong_idx[:12]
    show_predictions(X_test[pick], Y_test_true[pick], Y_test_pred_classes[pick], "Dự đoán sai (test)")
```
